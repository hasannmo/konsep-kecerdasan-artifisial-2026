DFS

```python
def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()
    while stack:
        node, path = stack.pop()
        if node == goal:
            return path
        if node in visited:
            continue
        visited.add(node)
        for neighbor, weight in reversed(graph[node]):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))
    return None

route = dfs(graph, "Arad", "Bucharest")
print("Route:", route)
```

BFS

```python
from collections import deque

def bfs(graph, start, goal):
    visited = {start}
    queue = deque([start])
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        if node == goal:
            return order
        for neighbor, weight in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return order

route = bfs(graph, "Arad", "Bucharest")
print("Route:", route)
```

UCS

```python
import heapq

def ucs(graph, start, goal):
    heap = [(0, start, [start])]
    visited = set()
    while heap:
        cost, node, path = heapq.heappop(heap)
        if node == goal:
            return path, cost
        if node in visited:
            continue
        visited.add(node)
        for neighbor, weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(heap, (cost + weight, neighbor, path + [neighbor]))
    return None, float("inf")

route, cost = ucs(graph, "Arad", "Bucharest")
print("Route:", route)
print("Total:", cost)
```
